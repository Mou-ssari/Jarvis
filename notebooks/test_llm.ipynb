```python
import time
from llama_cpp import Llama

llm = Llama(
    model_path=r"C:\Projects\Jarvis\models\qwen2.5-1.5b-instruct-q5_k_m.gguf",
    n_ctx=2048,
    n_threads=8,
    verbose=False
)

start = time.time()

response = llm.create_chat_completion(
    messages=[
        {"role": "system", "content": "You are JARVIS, a concise AI assistant."},
        {"role": "user", "content": "Introduce yourself in 1 sentence."}
    ],
    max_tokens=100,
    temperature=0.7
)

elapsed = time.time() - start

content = response["choices"][0]["message"]["content"]
tokens = response["usage"]["completion_tokens"]

print(f"Response: {content}")
print(f"Time: {elapsed:.2f}s")
print(f"Tokens generated: {tokens}")
print(f"Speed: {tokens / elapsed:.1f} tokens/sec")
```
