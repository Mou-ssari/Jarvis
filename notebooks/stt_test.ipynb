```python
import time
from faster_whisper import WhisperModel
```

```python
model = WhisperModel("tiny.en", device="cpu", compute_type="int8")

start = time.time()
segments, info = model.transcribe("C:/Projects/Jarvis/test.wav")
end = time.time() - start

text = " ".join([segment.text for segment in segments])
print(f"Detected language '{info.language}' in {end:.2f}s")
print(f"Transcription: {text}")
print(f"Time: {end:.2f}s")
print(f"Audio length: {info.duration:.2f}s")
print(f"Real-time factor: {end / info.duration:.2f}x realtime")
```
